```python
import json
with open("meta.json", "r") as f:
    meta_data = json.load(f)
keys_new = [f"{m['year']}_{m['president']}".replace(" ", "_") for m in meta_data.values()]

meta = {k:v for k,v in zip(keys_new, meta_data.values())}
print (meta)
with open("meta_new.json", "w") as f:
    json.dump(meta, f, indent=4)
```
