```python
import pygraphviz as pgv

# 1. Erstellen eines leeren Graphen
G = pgv.AGraph()

# 2. Hinzufügen von Knoten und Kanten
G.add_edge("A", "B")
G.add_edge("A", "C")
G.add_edge("B", "D")

# 3. Layout berechnen (z.B. mit dem 'dot'-Algorithmus)
G.layout(prog="sfdp")

# 4. In eine Datei zeichnen
G.draw("graph.svg", format="svg")
```
