```python
from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.dml.color import RGBColor
from pptx.enum.text import PP_ALIGN

# Initialize presentation
prs = Presentation()
blank_slide_layout = prs.slide_layouts[6] # Blank layout
slide = prs.slides.add_slide(blank_slide_layout)

# Define YlGnBu9 RGB Colors
ylgnbu9_rgb = [
    (255, 255, 217), (237, 248, 177), (199, 233, 180),
    (127, 205, 187), (65, 182, 196),  (29, 145, 192),
    (34, 94, 168),   (37, 52, 148),   (8, 29, 88)
]

# Title text box
tx_box = slide.shapes.add_textbox(Inches(0.5), Inches(0.5), Inches(9), Inches(1))
tf = tx_box.text_frame
p = tf.add_paragraph()
p.text = "Bokeh YlGnBu9 Color Palette"
p.font.size = Pt(32)
p.font.bold = True

# Create a continuous horizontal color bar / grid layout
box_width = Inches(1.0)
box_height = Inches(2.5)
start_left = Inches(0.5)
top_position = Inches(2.0)

for i, rgb in enumerate(ylgnbu9_rgb):
    # Add rectangle shape
    left_pos = start_left + (i * box_width)
    shape = slide.shapes.add_shape(
        1,  # 1 corresponds to a rectangle (MSO_SHAPE.RECTANGLE)
        left_pos, top_position, box_width, box_height
    )
    
    # Fill color
    shape.fill.solid()
    shape.fill.fore_color.rgb = RGBColor(*rgb)
    shape.line.color.rgb = RGBColor(*rgb) # Remove border line
    
    # Add a label underneath each box
    lbl_box = slide.shapes.add_textbox(left_pos, top_position + box_height + Inches(0.2), box_width, Inches(0.5))
    lbl_tf = lbl_box.text_frame
    lbl_p = lbl_tf.paragraphs[0]
    lbl_p.text = f"Color {i+1}"
    lbl_p.alignment = PP_ALIGN.CENTER
    lbl_p.font.size = Pt(11)

# Save the presentation file
prs.save("YlGnBu9_Palette.pptx")
print("PowerPoint palette saved as 'YlGnBu9_Palette.pptx' successfully!")
```

```python
import itertools
import networkx as nx


def gemeinsame_b_knoten_nach_attributen(graph):
    # 1. Knotenmengen dynamisch über die Attribute filtern
    nodes_A = [
        n for n, attr in graph.nodes(data=True) if attr.get("type") == "term"
    ]
    # (Optional, aber gut zur Sicherheit: Wir merken uns auch, wer vom Typ "address" ist)
    nodes_B_set = {
        n for n, attr in graph.nodes(data=True) if attr.get("type") == "address"
    }

    # 2. Dictionary für die Paare aus A erstellen
    ergebnis = {}

    for u, v in itertools.combinations(nodes_A, 2):
        # Schnittmenge aller gemeinsamen Nachbarn bilden
        gemeinsame = set(graph.neighbors(u)) & set(graph.neighbors(v))

        # Sicherstellen, dass die gemeinsamen Nachbarn auch wirklich vom Typ "address" (B) sind
        gemeinsame_B = gemeinsame & nodes_B_set

        # Nur hinzufügen, wenn die Liste nicht leer ist
        if gemeinsame_B:
            ergebnis[(u, v)] = list(gemeinsame_B)

    return ergebnis


# --- BEISPIEL-GRAPH ERSTELLEN ---
G = nx.Graph()

# Knoten mit Attributen hinzufügen
G.add_node("Term1", type="term")
G.add_node("Term2", type="term")
G.add_node("Term3", type="term")

G.add_node("Addr1", type="address")
G.add_node("Addr2", type="address")
G.add_node("Addr3", type="address")

# Kanten hinzufügen
G.add_edges_from(
    [
        ("Term1", "Addr1"),
        ("Term1", "Addr2"),
        ("Term2", "Addr1"),
        ("Term2", "Addr2"),  # Term1 und Term2 teilen Addr1 und Addr2
        ("Term3", "Addr3"),
    ]
)

# --- FUNKTION AUSFÜHREN ---
ergebnis_dict = gemeinsame_b_knoten_nach_attributen(G)
print(ergebnis_dict)
# Ausgabe: {('Term1', 'Term2'): ['Addr1', 'Addr2']} (Reihenfolge in der Liste kann variieren)
```
