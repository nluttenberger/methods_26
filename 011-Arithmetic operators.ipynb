### Recap: Integers & Floats

An *integer* and a *float* (short for *floating point number*) are two Python data types for representing numbers. Integers represent whole numbers. Floats represent numbers with decimal points. They do not need to be placed in quotation marks.

```python
type(40)
```

```python
type(40.5)
```

```python
type(40.555555)
```

### Arithmetic Operators

You can do a large range of mathematical calculations and operations with integers and floats. The table below is taken from Python's documentation about [Numeric Types](https://docs.python.org/3/library/stdtypes.html#numeric-types-int-float-complex).

| Operation       | Explanation                                                                      |
|-----------------|-----------------------------------------------------------------------------|
| `x` + `y`           | sum of `x` and `y`                                                              |
| `x` - `y`           | difference of `x` and `y`                                                       |
| `x` * `y`           | product of `x` and `y`                                                          |
| `x` / `y`           | quotient of ``x`` and `y`                                                         |
| ``x`` // `y`          | floored quotient of `x` and `y`                                                 |
| `x` % `y`           | remainder of `x` / `y`                                                          |
| -`x`              | `x` negated                                                                   |
| +`x`              | `x` unchanged                                                                 |
| `abs(x)`          | absolute value or magnitude of `x`                                         |
| `int(x)`          | `x` converted to integer                                                      |
| `float(x)`        | `x` converted to floating point                                               |
| `pow(x, y)`       | `x` to the power `y`                                                            |
| `x` ** `y`          | `x` to the power `y`                                                            |


Basic arithmetic operations can be performed on numbers, both integers and floats.

```python
a = 1
b = 2
print (a + b)
print (a - b)
print (a * b)
```

The division operator '/' always returns a float value, even if both operands are integers.


```python
print (a / b)
```

If we want to get the integer part of the division of the first operand by the second operand, we can use the '//' operator; also called floor division.


```python
print (a // b)
```

The exponentiation operator '**' raises the value of the first operand to the power of the second operand.

```python
print (a ** b)
```

The modulo operator '%' returns the remainder of the integer division of the first operand by the second operand.

```python
print (a % b)
```

We can increment the value of a variable by using the '+=' operator. 


```python
a = 1
```

The following instruction adds the value of the right operand to the value of the left operand and assigns the result to the left operand.

```python
a += 5
print(a)
```

Similarly, we can use the '-=', '*=', '/=', '//=', '**=', and '%=' operators to perform the corresponding operations on the variable.

Some arithmetic operations can also be performed on strings.
Concatenation of strings is done with the '+' operator.

```python
a = "Hello" 
b = " World!"
print(a + b)
```

The multiplication operator '*' can be used to repeat a string a certain number of times.

```python
print(a * 3)
```
