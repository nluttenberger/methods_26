### Types

Python has four collection types:
- `list` of items:
Items are enclosed in square brackets and separated by commas.
Lists can contain items of different types, including other lists.

```python
li = [1, 2.0, "Hello", True]
```

- `dictionary` of items:
Dictionaries are unordered collections of key-value pairs.
Dicts are mutable, which means we can change their items.
Dict key-value pairs are enclosed in curly braces and separated by commas.
Each key-value pair is separated by a colon.

```python
di = {"name": "Alice", "age": 30, "city": "New York"}
```

- `set` of items: 
Sets are unordered collections of unique items.
Sets are mutable, which means we can add or delete items to/from a set.
They are similar to lists, but they do not allow duplicate items.
We can create a set using the set() constructor or by enclosing items in curly braces.

```python
se = {1, 2, 3, 4, 5}
```

-  `tuple` of items:
Tuples are ordered collections of items that are immutable, meaning they cannot be changed after creation.
Tuples are order-preserving, which means the order of items is important.
Tuples items are separated by commas; they may be enclosed in parentheses.

```python
tu = ('XS', 'S', 'M', 'L', 'XL')
```
