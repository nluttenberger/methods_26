### Reading and writing

#### Read a text file

If you want to read or write a text file with Python, it is necessary to first open the file. To open a file, you can use Python's built-in `open()` function. The `read()` method returns the file contents.

```python
f = open("misc/sample-file.txt", mode="r", encoding="utf-8")
print(f.read())
f.close()
```

Inside the `open()` function parentheses, you insert the filepath to be opened in quotation marks. You should also insert a character encoding, which we will talk more about below. 

It is required to close a file after usage: `close()`.

It is good practice to **always** use the `with` keyword when dealing with file objects. The advantage is that the file is properly closed after its suite finishes, even if an exception is raised at some point.  

```python
with open("misc/sample-file.txt", "r", encoding='utf-8') as f:
  print(f.read())
```

#### Write text to a text file

To write text to a file, you need to use the `w` mode inside the open method and the `.write()` method. But be carefull: The `.write()` method overwrites all existing content!

```python
with open("misc/a-new-file.txt", "w", encoding='utf-8') as f:
  f.write("Now the file has some content! ")

#open and read the file after writing:
with open("misc/a-new-file.txt", "r", encoding='utf-8') as f:
  print(f.read())
```

#### Append text to a text file

If you don't want to overwrite the content of an existing file, but only append some new content to it, use the 'a' mode.

```python
with open("misc/a-new-file.txt", "a", encoding='utf-8') as f:
  f.write("Now the file has more content!")

#open and read the file after the appending:
with open("misc/a-new-file.txt", "r", encoding='utf-8') as f:
  print(f.read())
```

### Character Encoding

```python
encoding='utf-8'
```

Why do we need to include `encoding='utf-8'` to open our text file? Well, UTF-8 is a character encoding (a specific kind of [Unicode](https://home.unicode.org/basic-info/faq/)). We need to specify a character encoding because — *gasp!* — computers don't actually know what text is. Character encodings are systems that map characters to numbers. Each character is given a specific ID number. This way, computers can actually interpret strings of Ones and Zeros as characters.

You can check any characters' "code point" (i.e. place in the Unicode universe) with the function `ord()`

```python
ord("a")
```

```python
ord("💩")
```

```python
ord("ত")
```

```python
ord("!")
```

### Open and Read All Files in a Directory

It's helpful to know how to work with all the files in a directory.

#### Import Path library

```python
from pathlib import Path
```

```python
directory_path = 'misc'
```

#### Loop through any file in the directory

Use the star `*` character, which matches anything

```python
for filepath in Path(directory_path).glob('*'):
    print(filepath)
```

Loop through just text files in the directory with `*.txt`, which matches only files that end with ".txt"

```python
for filepath in Path(directory_path).glob('*.txt'):
    print(filepath)
```

To read these text files, simply add in the `with` construct with the `.read()` method. Instead of the `read()` method we here show the `readline()` method which reads only the first line of a text file.

```python
for filepath in Path(directory_path).glob('*.txt'):
    with open(filepath, "r", encoding='utf-8') as f:
        print(f.readline())
```
