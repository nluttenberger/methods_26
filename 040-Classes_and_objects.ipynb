## Classes and Objects

---

### Why do we use Classes?

Classes provide a means of bundling data and functionality together. 

### What is a Python Class ?

The Python `class` keyword creates a user-defined structure, which holds *class attributes* and *class methods*. 

From a `class`, instances of this class can be created. An instance of a class is called an `object`.

A `class`/an `object` consists of :

 - State : It is represented by attributes of an object, i.e. by a number of instance variables, see below. 
 - Behavior : It is represented by methods of an object. 
 - Identity : Each object has a unique name.

 
 (Image from: https://www.scholarhat.com/tutorial/python/python-classes-and-objects)

So you may say: 

When an object of a class is created, the class is said to be instantiated. A single class may have any number of instances. 

All the instances of a class share the behavior of the class. But the object state -- as reflected in the values of the instance variables -- is unique for each object. They are accessed inside a class method with `self` prepended to the variable name, e.g. `self.var1`.

### How do we create a Class?

Here are simple rules to create a class in Python:
 - Classes are created by keyword `class`.
 - Attributes are the variables that belong to class. 
 - Attributes are always public and can be accessed using dot (.) operator, e.g. `object.attribute`.
 - Methods are accessed the same way, i.e. via the dot operator, e.g. `object.method()`. Methods can have parameters, e.g. `object.method(param)`.

The **__init__** method is a constructor. Constructors are used to initialize the object’s state. The constructor is executed at time of Object creation. 
The method is useful to do any initialization you want to do with your object.

Inside the class structure, the first argument of a method is called `self`. By using the **self** keyword we can access the attributes and methods of the class. 

Here is an example:

```python
class Dog:  
    
    # instance variables
    # their names are shared by all instances of the class,
    # but their values are unique for each instance
    species = "pet"
    
    # constructor (a special method) 
    # adds further instance variables and initializes them
    def __init__(self, name, age):  
        self.name = name
        self.age = age
  
    # A sample method   
    def fun(self):  
        print("Bark, bark!") 

# Object instantiation 
Justin = Dog('Justin', 5) 
  
# Accessing class attributes and methods
print(Justin.name)
print(Justin.age)
print(Justin.species) 
Justin.fun() 
```
