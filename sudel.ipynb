```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd


corpus = [
    'Machine learning is great Machine',
    'Natural language processing and machine learning',
    'Language processing is part of NLP'
]

tfidf_vectorizer = TfidfVectorizer()
tfidf_vector = tfidf_vectorizer.fit_transform(corpus)
# Make a DataFrame out of the resulting tf–idf vector, setting the "feature names" (terms) as columns and the address titles (i.e. file names) as rows
tfidf_df = pd.DataFrame(tfidf_vector.toarray(), index=corpus, columns=tfidf_vectorizer.get_feature_names_out())
tfidf_df.loc['Total'] = tfidf_df.mask(tfidf_df > 0, 1).sum()
print (tfidf_df)

count_vectorizer = CountVectorizer(binary=True)
count_vector = count_vectorizer.fit_transform(corpus)
# Make a DataFrame out of the resulting count vector, setting the "feature names" (terms) as columns and the address titles (i.e. file names) as rows
count_df = pd.DataFrame(count_vector.toarray(), index=corpus, columns=count_vectorizer.get_feature_names_out())

count_df.loc['Total'] = count_df.sum()
#print (count_df)
```
