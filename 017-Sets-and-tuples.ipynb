## Sets and tuples

---

### Sets

Set values are unique. Set values are unordered. 

The `set()` function creates a set in Python.

```python
# Create empty set
a = set()
a
```

Create set from list

```python
a = set([1, 2, 3, 4, 2, 5])
a
```

Note: We cannot create empty sets using `{}` syntax as it creates an empty dictionary. To create an empty set, we use `set()` as shown above.

Values can be added/removed via the `add`/`remove` methods.

```python
a.add(6)
a.remove(2)
a
```

We can check, whether an element is in a set or not, by the `in` method.

```python
2 in a
```

We can perform different set operations.

```python
b = set(["a", "b", "c", "c", "d"])
a.union(b) 
```

```python
a.intersection(b)   
```

### Tuples

Tuple values need not be unique. Tuple values are ordered. The number of elements in a tuple is fixed, once it is created, and the items are unchangeable. 

Create a tuple

```python
sizes = ("XS", "S", "M", "L", "XL")
sizes
```

Access tuple items by index

```python
sizes[0]
```

Combine tuple items with their index

```python
sizes_indexed = list(enumerate(sizes))
sizes_indexed
```

Items of this list are of type tuple.

```python
type(sizes_indexed[0])
```

You may ask: Is M bigger then L?

```python
if sizes.index('M') > sizes.index('L'):
    print("M is bigger than L")
else:
    print("L is bigger than M") 
```
